## Setup and Dataset
First, let's load a real dataset of tweets using the NLTK library. Run this setup cell to construct your `corpus` variable containing 100 cleaned tweets.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import nltk

# Download NLTK Twitter sample data
nltk.download('twitter_samples', quiet=True)
from nltk.corpus import twitter_samples

# Fetch actual tweets (we will use 100 positive tweets for a clean, small sample size)
tweets = twitter_samples.strings('positive_tweets.json')[:100]

# Clean up some basic twitter artifacts (like URLs and handles) for better NLP results
import re
def clean_tweet(tweet):
    tweet = re.sub(r'http\S+|www\S+|https\S+', '', tweet, flags=re.MULTILINE) # remove URLs
    tweet = re.sub(r'@\w+', '', tweet) # remove handles
    return tweet.strip()

corpus = [clean_tweet(t) for t in tweets if len(clean_tweet(t)) > 10]

print(f"Loaded {len(corpus)} real tweets into 'corpus'.")
print("\nSample Tweets:")
for idx, doc in enumerate(corpus[:5]):
    print(f"Tweet {idx + 1}: {doc}")
```

## Task 3: TF-IDF Representation
TF-IDF (Term Frequency-Inverse Document Frequency) down-weights common words that appear across many documents, highlighting terms that are uniquely important to individual documents.

**Instructions:**
1. Initialize `TfidfVectorizer` and fit/transform your `corpus` to get the TF-IDF representation matrix.
2. Create a Pandas DataFrame with the features as column headers.
3. Display the first 5 rows of the DataFrame, rounding values to 3 decimal places.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Initialize TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# 2. Fit and transform the 'corpus'
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# 3. Create a DataFrame and display its head rounded to 3 decimal places
feature_names = tfidf_vectorizer.get_feature_names_out()
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

display(tfidf_df.head().round(3))
```
